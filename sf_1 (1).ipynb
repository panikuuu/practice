Когда нужно сгруппировать данные по одному или нескольким категориальным столбцам, а показать агрегировнные значения по другим столбцам, то используются фнукция .groupby(). Ниже представлена группировка по позиции, и для сгруппированных данных для пересечения, финиша и силе удара найдено максимальное значение

```
data.groupby(['Position'])[['Crossing','Finishing','ShotPower']].max()
```

value_counts() - считает количество строк для каждого значения категориальной переменной. Можно присвоить результаты выборки некоторой переменной и проводить манипуляции с ними

```
dd = data['Position'].value_counts()
```

```
dd
```

```
dd.shape
```

```
dd.loc[dd > 500]
```

value_counts() - можно использовать и для числовой, не категориальной перменной, и в таком случае лучше добавить число интервалов bins, на которой нужн разделить отрезок, иначе каждое значение будет считаться уникальным и это часто не слишком помагает анализу

```
data['SprintSpeed'].value_counts(bins = 6)
```

sort_values(ascending=False) - позволяет сортировать данные, при ascending=False даныые сортируются по убыванию, при ascending=True по возрастанию

```
data.groupby(['Position'])['Wage'].max().sort_values(ascending=False)
```

```
# Средний возраст по клубам
ss = data.groupby(['Club'])['Age'].mean()
ss.plot()
```

```
# Средний возраст по клубам
ss = data.groupby(['Club'])['Age'].mean().sort_values(ascending=False)
ss
```

```
ss.plot()
```

```
#Средняя ловкость по клубам
aa =data.groupby(['Club'])['Agility'].mean().sort_values(ascending=False)
aa.plot()
```

```
#Как меняется средний баланс в зависимости от позиции
a = data.groupby(['Position'])['Balance'].mean().sort_values(ascending=False)
a
```

```
#Как меняется средний баланс в зависимости от позиции график
a.plot(kind = 'bar')
```

```
#По сколько игроков на каждой позиции график
b = data['Position'].value_counts()
b.plot(kind = 'bar')
```

```
#Сколько пенальти у каждого клуба
data.groupby(['Club'])['Penalties'].sum().sort_values(ascending=False)
```
